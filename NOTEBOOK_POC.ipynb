# DarkPyonix 스트림 핸드오프 PoC

이 노트북은 커널 매니저가 SSE 엔드포인트를 열고 커널이 제어 채널로 attach_stream 명령을 받아 직접 이벤트를 push 하는 흐름을 설명합니다.

```python
import httpx, asyncio, json
async def demo():
    async with httpx.AsyncClient(base_url='http://localhost:8000') as client:
        r = await client.post('/kernels/poc')
        print('kernel created', r.json())
        await asyncio.sleep(0.4)
        r = await client.get('/kernels/poc/stream')
        print('stream id', r.headers.get('X-Stream-ID'))
        async for b in r.aiter_bytes():
            print(b.decode().strip())
            if 'i' in b.decode():
                break
asyncio.run(demo())
```
